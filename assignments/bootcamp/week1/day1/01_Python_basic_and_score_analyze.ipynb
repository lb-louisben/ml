1. What is f-strings?

f-string 是“格式化字符串”，用于把变量、计算结果插入字符串。

the basic structure

- f"text {vars or expressions} "

`.2f` is a format specifier used in programming (most notably in Python) to control how floating-point numbers are displayed.

It forces a number to be formatted as a string with exactly two decimal places, automatically applying rounding if necessary.


```python
stu_programming_score = 80.5
stu_math_score = 90.5
stu_avr = (stu_programming_score + stu_math_score) / 2
print(f"Average Score: {stu_avr:.1f}")
```

```python
score = 85.5678

print(f"{score:.0f}")
print(f"{score:.1f}")
print(f"{score:.2f}")
print(f"{score:.3f}")
print(f"{score}")


# Example 1: Rounding down
pi = 3.1415926
print(f"Pi: {pi:.2f}")  
# Output: Pi: 3.14

# Example 2: Rounding up
price = 12.996
print(f"Price: ${price:.2f}")  
# Output: Price: $13.00

# Example 3: Padding with trailing zeros
cash = 5.5
print("Balance: ${:.2f}".format(cash))  
# Output: Balance: $5.50
```

## LOOP & FUNCTION

任务：编写 calculate_pass_rate() 函数。

```python
scores = [72, 85, 91, 56, 78]
pass_scores = [score for score in scores if score >= 60]

print(f"通过的分数: {pass_scores}")
print(f"通过率: {len(pass_scores) / len(scores) * 100:.2f}%")
adjusted_scores = [min(score + 5, 100) for score in scores]
print(f"调整后的分数: {adjusted_scores}")
print(f"平均分数: {sum(adjusted_scores) / len(adjusted_scores):.2f}")
```

```python
def calculate_pass_rate(scores):
    # 在这里完成代码
    pass_count = sum(1 for score in scores if score >= 60)
    return pass_count / len(scores) if scores else 0

scores = [72, 85, 91, 56, 78]
print(calculate_pass_rate(scores))
```

## DAY 1 ASSIGNMENT

original data as followed: 


```python
students = [
    {"name": "小明", "score": 85},
    {"name": "小红", "score": 92},
    {"name": "小刚", "score": 58},
    {"name": "小丽", "score": 76},
    {"name": "小王", "score": 64}
]
```

##### 任务 1：输出所有学生的信息

- 小明：85分，良好
- 小红：92分，优秀
- 小刚：58分，不及格

```python
for student in students:
    score = student["score"]
    if score < 60:
        result = "不及格"
    elif 60 <= score <= 90:
        result = "良好"
    else:
        result = "优秀"
    student["result"] = result

print("\n".join([f"{i['name']}: {i['score']}分，{i['result']}" for i in students]))

print()
print("---------- standard solution ----------")
print()

# stadnard solution
for student in students:
    score = student["score"]
    if score < 60:
        result = "不及格"
    elif 60 <= score <= 90:
        result = "良好"
    else:
        result = "优秀"
    student["result"] = result
    
    print(f"{student['name']}: {student['score']}分，{student['result']}")
```

##### 任务 2：计算班级统计数据
需要计算：
- 学生人数
- 平均分
- 最高分
- 最低分
- 及格人数
- 及格率

```python
student_amount = len(students)
print(f"student amount: {student_amount}")
print(f"average score: {sum(student['score'] for student in students) / student_amount}")
print(f"the highest score: {max(student['score'] for student in students)}")
print(f"the lowest score: {min(student['score'] for student in students)}")
print(f"passed_count: {sum(1 for student in students if student['score'] >= 60)}")
print(f"passed_rate: {sum(1 for student in students if student['score'] >= 60) / student_amount * 100:.2f}%")
```

##### 任务3: 找到最高分学生

```python
top_student = max(
    students,
    key=lambda student: student["score"]
)

print(
    f"最高分学生：{top_student['name']}，"
    f"成绩：{top_student['score']}"
)
```
