### 1. 基础回顾：变量、列表、字典与函数

字符串保存文本；布尔值保存真假；列表保存序列；字典保存键值对应。
你曾遇到 {"passed"} 的输出：它是一个集合；"passed" 才是字符串。
本练习成绩满分 100，及格线 60。第三天真实数据采用 20 分制，不要混用。

```python
students = [{"name": "小明", "score": 85}, {"name": "小红", "score": 92},
            {"name": "小刚", "score": 58}, {"name": "小丽", "score": 76},
            {"name": "小王", "score": 64}]
scores = [s["score"] for s in students]
print(f"平均分：{sum(scores)/len(scores):.1f}")
print(type("passed"), type({"passed"}), type({"passed": True}))
```

### 2. 深度：边界条件和空列表

等级定义：90～100 优秀，80～不足90 良好，60～不足80 及格，小于60不及格。
原笔记中的 60～90 都归为良好，与这套规则不同，保留原笔记供对照。
空列表的及格率没有分母，应明确返回 None 或抛出异常；返回 0 会与“所有人都不及格”混淆。

```python
def get_grade(score):
    if not 0 <= score <= 100:
        raise ValueError("成绩必须在 0～100")
    if score >= 90:
        return "优秀"
    if score >= 80:
        return "良好"
    if score >= 60:
        return "及格"
    return "不及格"

def calculate_pass_rate(scores):
    if not scores:
        return None
    return sum(score >= 60 for score in scores) / len(scores)

for score in [0, 59, 60, 79, 80, 89, 90, 100]:
    print(score, get_grade(score))
print("空列表：", calculate_pass_rate([]))
print(f"及格率：{calculate_pass_rate(scores):.1%}")
```

### 3. 深度：可变对象、引用与复制

b=a 让两个变量指向同一个列表。列表的 copy 仅复制外层；嵌套字典仍可能共享。
编写函数时尽量避免无意修改调用者传入的数据。

```python
a = [1, [2,19]]

# 引用传递，赋值
b = a
b.append(3)
print(f"引用传递：a 和 b 共享同一个列表对象，修改 b 会影响 a")
print(f"原始列表：{a}, 共享列表：{b}")
print(f"原始列表 id：{id(a)}, 共享列表 id：{id(b)}")
# print(f"原始列表地址：{hex(id(a))}, 共享列表地址：{hex(id(b))}")
print(f"原始列表地址：{id(a):#x}, 共享列表地址：{id(b):#x}")

print()
print()

# 浅拷贝：a 和 c 是两个独立的列表对象，修改 c 不会影响 a
c = a.copy()
# 改了 c 的结构与内容
c.append(4)
print(f"浅拷贝：a 和 c 是两个独立的列表对象，修改 c 不会影响 a")
print(f"原始列表：{a}, 独立列表：{c}")
print(f"原始列表 id：{id(a)}, 独立列表 id：{id(c)}")
print(f"原始列表第一位元素地址：{id(a[1]):#x}, 独立列表地址：{id(c[1]):#x}")
# print(f"原始列表地址：{hex(id(a))}, 独立列表地址：{hex(id(c))}")
print(f"原始列表地址：{id(a):#x}, 独立列表地址：{id(c):#x}")

print()
print()

d = a.copy()
d[1].append(99)
print(f"浅拷贝：a 和 d 是两个独立的列表对象，但 d 中的元素是共享的，修改 d 中的元素会影响 a")
print(f"原始列表：{a}, 独立列表：{d}")
print(f"原始列表 id：{id(a)}, 独立列表 id：{id(d)}")
print(f"原始列表地址：{id(a):#x}, 独立列表地址：{id(d):#x}")

print()
print()

# 深拷贝：a 和 d 是两个独立的列表对象，且 d 中的元素也是独立的对象，修改 d 不会影响 a
import copy
independent_students = copy.deepcopy(students)
independent_students[0]["score"] = 0
assert students[0]["score"] == 85

print("原始学生列表：", students)
print("独立学生列表：", independent_students)
print(f"原始学生列表 id：{id(students)}, 独立学生列表 id：{id(independent_students)}")
print(f"原始学生列表地址：{id(students):#x}, 独立学生列表地址：{id(independent_students):#x}")
```

### 4. 广度：enumerate、zip、排序与错误阅读

enumerate 提供序号；zip 配对两组元素，默认在最短输入结束。
sorted 返回新列表。KeyError 检查字典键，NameError 检查定义顺序，TypeError 检查类型。
try/except 应处理预期错误，不能隐藏所有失败。

```python
# enumerate() 函数可以将可迭代对象组合为一个索引序列，同时列出数据和数据下标
for rank, student in enumerate(sorted(students, key=lambda s: s["score"], reverse=True), 1):
    print(rank, student["name"], student["score"])

# zip() 函数可以将两个可迭代对象打包成一个元组列表
for name, score in zip(["A", "B"], [70, 80]):
    print(name, score)

# try-except 语句可以捕获异常，避免程序崩溃
try:
    get_grade(105)
except ValueError as error:
    print("已捕获预期错误：", error)
```

### DAY 1 ASSIGNMENT

- 基础：编写 analyze_students，返回人数、均分、最高分、最低分和及格率。
- 深度：规定空班级怎样处理，并测试 60/80/90 分边界。
- 广度：用 sorted 输出前三名，解释同分时的处理。
- Bonus：返回所有并列最高分学生，而非只返回第一个；证明函数没有修改输入。
- 验收：本例人数 5，均分 75，最高 92，最低 58，及格率 80%。

```python
# 在这里完成函数。
def analyze_students(students):
    # 空班级
    if not students:
        raise ValueError("班级中没有学生")
    students_size = len(students)
    scores = [s["score"] for s in students]
    average_score = sum(scores) / students_size

    max_score = max(scores)
    min_score = min(scores)
    pass_rate = calculate_pass_rate(scores)
    return {
        "students_size": students_size,
        "average_score": average_score,
        "max_score": max_score,
        "min_score": min_score,
        "pass_rate": pass_rate
    }

def the_top_3_students(students):
    # 空班级
    if not students:
        raise ValueError("班级中没有学生")
    # 解释同分数处理
    for rank, student in enumerate(sorted(students, key=lambda s: s["score"], reverse=True)[:3], 1):
        print(rank, student["name"], student["score"])

analyze_students(students)
the_top_3_students(students)


```

### 参考答案

```python
def analyze_students(students):
    if not students:
        return {"count": 0, "mean": None, "max": None, "min": None,
                "pass_rate": None, "top_names": []}
    values = [s["score"] for s in students]
    for value in values:
        get_grade(value)
    maximum = max(values)
    return {"count": len(values), "mean": sum(values)/len(values),
            "max": maximum, "min": min(values), "pass_rate": calculate_pass_rate(values),
            "top_names": [s["name"] for s in students if s["score"] == maximum]}
print(analyze_students(students))

assert analyze_students(students)["mean"] == 75
assert get_grade(90) == "优秀"
assert analyze_students([])["mean"] is None

print(analyze_students([{"name":"A","score":90},{"name":"B","score":90}]))
```
