### 1. list to array

```python
import numpy as np

scores_list = [72, 85, 91, 56, 78]
scores = np.array(scores_list)

print("Scores:", scores)
print("type of scores:", type(scores))

# compare the calculate of list and numpy array
print("list multiplication:", scores_list * 2)
print("numpy array multiplication:", scores * 2)

# 数量对应位置相乘
print("numpy * numpy:", scores * scores)
# print("list * list:", scores_list * scores_list)  # This will raise an error


print(scores + 5)     # 每个人加 5 分
print(scores >= 60)   # 分别判断是否及格
```

```python
adjusted_scores = scores +3
print("Adjusted Scores:", adjusted_scores)
```

### 2. 2D array and shape

```python
names = np.array(["Louis", "Alice", "Bob", "Cindy", "David"])
subjects = np.array(["Python", "Math", "English"])

scores = np.array([
    [85, 78, 92],
    [90, 88, 95],
    [56, 65, 60],
    [78, 82, 80],
    [92, 91, 89]
])

print(scores)
print("形状：", scores.shape)
print("维度数量：", scores.ndim)
print("元素总数：", scores.size)
print("元素类型：", scores.dtype)

python_passed = scores[:, 0] >= 60
print(f"Python及格情况：{names[python_passed]}")
```

| 属性 | 本例结果 | 含义 |
|---|---|---|
| `shape` | `(5, 3)` | 5 行、3 列 |
| `ndim` | `2` | 有两个轴 |
| `size` | `15` | 共 15 个成绩 |
| `dtype` | 整数类型 | 数组元素的数据类型 |

`scores.shape[0]` 和 `scores.shape[1]` 分别是什么。

```python
print(scores[0]) # 第一行
```

### 3. index and slice

记住：
- `:` 表示这一轴全部取出。
- `1:3` 包含索引 `1`、`2`，不包含 `3`。

```python
print(scores[0, 0])    # Louis 的 Python 成绩：85
print(scores[0, :])    # Louis 的全部成绩
print(scores[:, 0])    # 全班的 Python 成绩
print(scores[-1, :])   # 最后一位学生的全部成绩
print(scores[:2, :])   # 前两位学生的全部成绩
print(scores[:, 1:3])  # 全班的 Math 和 English 成绩
```

```python
# 1. 取出 Bob 的 English 成绩。
print(f"Bob 的 English 成绩：{scores[2, 2]}")
# 2. 取出全班的 Math 成绩。
print(scores[:,1])
# 3. 取出最后两位学生的成绩。
print(scores[-2:,:])
# 4. 取出前三位学生的前两门成绩。
print(scores[:3,:2])
```

### 4. 按轴统计

```python
print("每位学生平均分：", scores.mean(axis=1))
print("每门课程平均分：", scores.mean(axis=0))
print("全班全部成绩平均分：", scores.mean())

student_averages = scores.mean(axis=1)
student_totals = scores.sum(axis=1)
subject_averages = scores.mean(axis=0)

print("每人平均分：", np.round(student_averages, 2))
print("每人总分：", student_totals)
print("各科平均分：", subject_averages)
print("各科最高分：", scores.max(axis=0))
print("各科最低分：", scores.min(axis=0))
```

练习：
1. 计算每位学生自己的最高分。
2. 计算每门课程的标准差：使用 `scores.std(axis=0)`。
3. 解释为什么计算“各科平均分”使用 `axis=0`。

标准差描述数据的分散程度：标准差越大，说明成绩越分散。`NumPy` 的 `std()`默认使用总体标准差算法。

```python
# 每位学生的最高分
print(scores.max(axis=1))

# 每门课程的标准差
print("各科标准差：", scores.std(axis=0))
```

### 5. 布尔筛选

第一天用 `for` 和 `if` 筛选，今天可以直接用`布尔数组`筛选。

```python
student_averages = scores.mean(axis=1)

mask = student_averages >= 80

print(mask)
print(names[mask])
print(scores[mask])
```

`mask` 相当于一张选择清单：`True` 对应的学生留下，`False` 对应的学生不选。

```python
# find the students who passed all subjects
all_passed = (scores >= 60).all(axis=1)
print(names[all_passed])

# 找出至少一门不及格的学生
at_least_one_failed = (scores < 60).any(axis=1)
print(names[at_least_one_failed])

# 组合条件
# Python 不低于 80，并且 Math 不低于 85
mask = (scores[:, 0] >= 80) & (scores[:, 1] >= 85)

print(names[mask])
```

数组中的逐元素条件组合用：
- `&`：`并且`
- `|`：`或者`
- `~`：`取反`

每个比较条件都用括号包起来。

```python
# 练习：找出平均分不低于 80、但 Math 低于 85 的学生。

mask = (student_averages >= 80) & (scores[:, 1] < 85)
print(names[mask])
```

### 6. 广播与矩阵乘法

广播让不同形状、但符合规则的数据进行运算。

```python
# 给每个成绩加 5 分
adjusted_scores = scores + 5
print(adjusted_scores)

# 分别给 Python、Math、English 加 2、5、0 分
bonuses = np.array([2, 5, 0])

adjusted_scores = scores + bonuses
print(adjusted_scores)
```
