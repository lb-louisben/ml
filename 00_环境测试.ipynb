# 机器学习环境测试

依次运行下面的代码单元，确认 Jupyter 和常用机器学习库工作正常。

```python
import sys
import numpy as np
import pandas as pd
import matplotlib
import seaborn as sns
import sklearn

print("Python:", sys.version.split()[0])
print("NumPy:", np.__version__)
print("Pandas:", pd.__version__)
print("Matplotlib:", matplotlib.__version__)
print("Seaborn:", sns.__version__)
print("scikit-learn:", sklearn.__version__)
```

```python
scores = pd.DataFrame({
    "study_hours": [1, 2, 3, 4],
    "score": [2, 4, 6, 8],
})
scores
```

```python
from sklearn.linear_model import LinearRegression

X = scores[["study_hours"]]
y = scores["score"]
model = LinearRegression().fit(X, y)
prediction = model.predict(pd.DataFrame({"study_hours": [5]}))[0]
print(f"学习 5 小时时的预测值：{prediction:.1f}")
```
